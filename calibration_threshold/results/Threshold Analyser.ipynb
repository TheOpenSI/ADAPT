```python
best_per_k = df.loc[
    df.groupby("k")["accuracy"].idxmax()
][[
    "k",
    "threshold",
    "accuracy",
    "general_recall",
    "specialist_recall",
    "fallback_rate"
]]

best_per_k
```

```python
global_threshold = (
    df[df["k"] >= 5]
#     df
    .groupby("threshold")[
        [
            "accuracy",
            "general_recall",
            "specialist_recall",
            "fallback_rate",
        ]
    ]
    .mean()
    .round(4)
)
```
