```python
import os
import json
import pandas as pd

base_path = "../results"

folders = {
    3: "dev_service_3",
    5: "dev_service_5",
    8: "dev_service_8",
    10: "dev_service_10"
}
```

```python
all_rows = []

for k, folder_name in folders.items():

    folder_path = os.path.join(base_path, folder_name)

    for file_name in os.listdir(folder_path):

        if not file_name.endswith(".json"):
            continue
            
        if "_efficiency" in file_name:
            continue

        file_path = os.path.join(folder_path, file_name)

        with open(file_path, "r") as f:
            data = json.load(f)

        for row in data:
            row["source_file"] = file_name
            all_rows.append(row)
```

```python
df = pd.DataFrame(all_rows)
```

```python
import numpy as np

def convert_to_softmax(score_dict):
    services = list(score_dict.keys())
    scores = np.array(list(score_dict.values()))

    # softmax
    exp_scores = np.exp(scores - np.max(scores))
    probabilities = exp_scores / exp_scores.sum()

    return dict(zip(services, probabilities))


qwen_8b_3["softmax_scores"] = qwen_8b_3["scores"].apply(convert_to_softmax)
```

```python
def get_true_probability(row):

    if row["expected_service"] == "general":
        return np.nan

    probability_column = "p_" + row["expected_service"]

    return row[probability_column]
```

```python
qwen_8b_3["true_class_probability"] = qwen_8b_3.apply(
    get_true_probability,
    axis=1
)
```

```python
qwen_8b_3["nonconformity"] = (
    1 - qwen_8b_3["true_class_probability"]
)
```

```python
import math

alpha = 0.10

# Only specialist examples are used for conformal calibration
calibration = qwen_8b_3[
    qwen_8b_3["expected_service"] != "general"
].copy()

# Get the 15 nonconformity scores and sort them
scores = sorted(calibration["nonconformity"].tolist())

n = len(scores)

# Conformal rank
position = math.ceil((n + 1) * (1 - alpha))

# Get conformal cutoff
q_hat = scores[position - 1]

print("Number of calibration examples:", n)
print("Position:", position)
print("Conformal cutoff:", q_hat)
print("Equivalent probability cutoff:", 1 - q_hat)
```

```python
service_names = [
    "abstract_algebra",
    "anatomy",
    "astronomy"
]

def get_conformal_set(row):

    prediction_set = []

    for service in service_names:

        probability = row["p_" + service]

        nonconformity = 1 - probability

        if nonconformity <= q_hat:
            prediction_set.append(service)

    return prediction_set
```

```python
qwen_8b_3["conformal_set"] = qwen_8b_3.apply(
    get_conformal_set,
    axis=1
)
```

```python
def get_top_scores(score_dict):
    
    services = list(score_dict.keys())
    scores = list(score_dict.values())
    
    sorted_scores = sorted(scores, reverse=True)
    
    top1_score = sorted_scores[0]
    top2_score = sorted_scores[1]
    
    top1_service = services[scores.index(top1_score)]
    top2_service = services[scores.index(top2_score)]
    
    margin = top1_score - top2_score
    
    return top1_service, top1_score, top2_service, top2_score, margin
```

```python
result = qwen_8b_3["scores"].apply(get_top_scores)
```

```python
qwen_8b_3[
    ["top1_service", "top1_score",
     "top2_service", "top2_score",
     "margin"]
] = pd.DataFrame(result.tolist(), index=qwen_8b_3.index)
```

```python
def get_gate_nonconformity(row):

    probabilities = row["softmax_scores"]

    # For a real specialist query:
    if row["expected_service"] != "general":
        true_probability = probabilities[row["expected_service"]]
        return 1 - true_probability

    # For General:
    highest_probability = max(probabilities.values())
    return 1 - highest_probability
```

```python
def get_binary_gate_nonconformity(row):

    probabilities = row["softmax_scores"]

    highest_probability = max(probabilities.values())

    return 1 - highest_probability
```

```python
import numpy as np

def get_gate_signals(row):

    probabilities = list(row["softmax_scores"].values())

    sorted_probabilities = sorted(
        probabilities,
        reverse=True
    )

    top1 = sorted_probabilities[0]
    top2 = sorted_probabilities[1]

    max_probability = top1
    margin = top1 - top2

    entropy = 0

    for p in probabilities:
        if p > 0:
            entropy = entropy - (p * np.log(p))

    return max_probability, margin, entropy
```

```python
results = qwen_8b_8.apply(
    get_gate_signals,
    axis=1
)

qwen_8b_8[
    ["max_probability", "margin", "entropy"]
] = pd.DataFrame(
    results.tolist(),
    index=qwen_8b_8.index
)
```

```python
def get_binary_label(service):

    if service == "general":
        return "general"

    return "specialist"


qwen_8b_8["binary_expected"] = qwen_8b_8[
    "expected_service"
].apply(get_binary_label)
```
